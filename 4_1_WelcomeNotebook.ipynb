## Rich, interactive outputs
Until now all of the generated outputs have been text, but they can be more interesting like the chart below.

Don't worry about what all the code means at this point, although the comments in the code cell (text preceded by a #) should give you an idea. You will learn this later on in the module.

````python
# import libraries
import numpy as np
from matplotlib import pyplot as plt

# generate random data
# np.random.randn(100) generates 100 random numbers from a standard normal distribution (mean = 0, standard deviation = 1).
# 200 + shifts the mean of the data to 200, so ys becomes a list of 100 random numbers centered around 200.
ys = 200 + np.random.randn(100)
x = [x for x in range(len(ys))] # create x-axis data

# plot the data
# this line creates a line plot of ys against x
plt.plot(x, ys, '-')
plt.fill_between(x, ys, 195, where=(ys > 195), facecolor='g', alpha=0.6) # fill the area under the curve

# add a title and show the plot
plt.title("Fills and Alpha Example")
plt.show()
````

## Integration with Google Drive

Colaboratory is integrated with Google Drive. It allows you to share, comment, and collaborate on the same document with multiple people:

* The **SHARE** button (top-right of the toolbar) allows you to share the notebook and control permissions set on it.

* **File->Save a Copy in Drive** creates a *new copy* of the notebook and saves it in your Google Drive, specifically in the My Drive/Colab Notebooks folder.

* **File->Save** saves the File to Drive.

* **File->Save and pin revision** pins the version so it doesn't get deleted from the revision history.

* **File->Revision history** shows the notebook's revision history.

It's important that you know the difference between **Save a Copy in Drive** and **Save**:

### 1. **File → Save a Copy in Drive**
   - **What it does**: This option creates a **new copy** of the current notebook and saves it in your Google Drive, specifically in the `My Drive/Colab Notebooks` folder.
   - **When to use**: Use this option when you want to **duplicate** the current notebook, creating a separate version that you can work on independently, without affecting the original file.
   - **Result**: You end up with two versions of the notebook: the original and the copy saved in your Drive.

### 2. **File → Save**
   - **What it does**: This option saves the current notebook you are working on, **overwriting** the existing version of the file (if it’s already in your Google Drive).
   - **When to use**: Use this option when you are working on a notebook already stored in your Google Drive and you want to **save changes** to it.
   - **Result**: The same notebook is updated with the latest changes, without creating any new copies.

### Key Difference:
- **"Save a Copy in Drive"** creates a duplicate, useful for version control or branching off into a new project.
- **"Save"** simply saves your progress on the current notebook, without creating a duplicate.
